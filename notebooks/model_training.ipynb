# Vehicle damage model – training & evaluation
Put your Kaggle / custom images and YOLO labels in `data/dataset/` first (see `damage.yaml`).

```python
# !pip install ultralytics matplotlib pandas
from ultralytics import YOLO
from pathlib import Path
ROOT = Path('..').resolve()
DATA = ROOT/'data'/'dataset'/'damage.yaml'
```

```python
model = YOLO('yolov8n.pt')
model.train(data=str(DATA), epochs=50, imgsz=640, project=str(ROOT/'runs'), name='damage')
```

```python
import shutil
shutil.copy(model.trainer.best, ROOT/'backend'/'models'/'damage_model.pt')
```

```python
m = model.val(data=str(DATA)).box
p, r = float(m.mp), float(m.mr)
print('Precision', p, 'Recall', r, 'F1', 2*p*r/max(p+r,1e-9), 'mAP50', m.map50)
```

```python
import pandas as pd, matplotlib.pyplot as plt
df = pd.read_csv(ROOT/'runs'/'damage'/'results.csv'); df.columns = df.columns.str.strip()
df.plot(x='epoch', y=['metrics/precision(B)','metrics/recall(B)','metrics/mAP50(B)'])
plt.title('Training metrics'); plt.show()
```
